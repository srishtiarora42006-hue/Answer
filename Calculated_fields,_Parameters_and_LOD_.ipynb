**Question 1** : What is a Calculated Field and why is it important in business dashboards?

-> A calculated field is used to add a special field to the existing dataset fields to get a new insight from the existing values. A calculated field lets a person calculate a new aspect from the present fields in the dataset, it is an important practice in building dashboards as it helps to analyse variables which may not be directly present and need to be derived from the existing columns ,dashboards show collective visualisations at one place these visualisations present relationships between diffrent measures and dimensions which hence provide meaning insights and hence are important for management for decision making and further planning.   

**Question 2** : Why can’t parameters change visuals directly in Tableau?

-> Parameters can't change the visuals directly in tableau because it is just a user defined single value and is not connected to any data or source , they are used with filters , calculated fields or other feautres to affect tableau visualisations.

**Question 3** : When should EXCLUDE LOD be used?

-> Exclude LOD is used when we want a level of detail to be ignored or not be concentrated upon, this usually is done when  we need to perform an aggregation using a higher level of detail or heirarchy , for eg. if we have calculated avg of sales by category by using exclude lod of products , then the visualisation will surely show the average of sales by category ignoring the product dimension , in this way the category is more into focus instead of product.

**Question 4**: Difference between FIXED and INCLUDE LOD expressions.

-> **FIXED LOD** - fixed LOD retains a dimension in the visual irrespective of what measure or dimension is being used there, this is generally used when there is need of a specific constant component in the view, it is independent of the dimensions used in the visuals.

**INCLUDE LOD** - include LOD is used when we need an extra level of detail in the view, despite the particular measure or dimension being used in the visual or not , it usually depends on the measure used in the visualisation and is generally used when the data needs to be represented at a higher level of heirachy but retention of a particular dimension is also necessary.
