**Q3.** Check if the price data is skewed.


```python
import matplotlib.pyplot as plt
plt.hist(df["Price"])
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.ticklabel_format(style='plain',axis = "x")
plt.xticks(rotation = 45)
plt.show()
```

This shows that the prices of the houses are **right skewed** means more number of people live in cheaper houses.  

**Q4**. What is the average price for each number of bedrooms?

```python
df["AVG_bedroom_price"] = df.groupby("number of bedrooms")["Price"].transform("mean")
print(df[["number of bedrooms","Price","AVG_bedroom_price"]])
```

**Q5.**  What is the relationship between living area and price? (Use : Area without Basement).

```python
# to find the relationship between living area and price we will perform bivariate analysis.
plt.scatter(df["Area of the house(excluding basement)"],df["Price"])
plt.xlabel("Area of the house(excluding basement)")
plt.ylabel("Price")
plt.show()

```

From the above scatter plot it is clear that **there is a positive relationship** between area of the house(excluding basement) and its price, as the area increases the price also increases.

**Q6.** Identify any anomalies where houses have high prices but low areas.

```python
# anomalies can be identified by applying this condition, these conditions are just threshold values chosen according to the dataset.
print(df.loc[
    (df["Price"] > 1000000) &
    (df["Area of the house(excluding basement)"] < 2000),
    ["Price", "Area of the house(excluding basement)"]
])
```
