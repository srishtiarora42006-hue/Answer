**Q1.** Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the dataset information to understand its structure.



```python
import pandas as pd


df = pd.read_excel("/content/retail_dataset .xlsx")

print('First 5 rows:')
display(df.head())
print('Dataset information:')
display(df.info())

```

**Q2.** Inspect the dataset and identify missing values in each column. Display the total number of missing
values present in the dataset.

```python
#missing values in each column
missing_values = df.isnull().sum()
print('Missing values in each column:')
print(missing_values)
```

**Q3.** Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())
```

**Q4.** Remove all rows where the Product Category or Region column contains missing values.


```python
df.dropna(subset=["Product Category", "Region"], inplace=True)
```

**Q5.** Create a new column called Revenue using NumPy, where:

Revenue = Quantity × Price

Display the updated dataset with the new column.

```python
import numpy as np
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])
print('Updated dataset with the new column:')
display(df.head())
```

**Q6.** Using NumPy, calculate the total revenue generated from all transactions in the dataset.

```python
Total_Revenue = np.sum(df["Revenue"])
print('Total revenue generated from all transactions in the dataset:')
print(Total_Revenue)
```

**Q7.** Group the dataset by Product Category and calculate the total revenue generated by each category.

```python
df.groupby("Product Category")["Revenue"].sum()
```

**Q8.** Based on the revenue calculated for each product category, identify the top 3 categories with the
highest revenue and the bottom 3 categories with the lowest revenue.

```python
category_revenue = df.groupby("Product Category")["Revenue"].sum()

top_3 = category_revenue.nlargest(3)
bottom_3 = category_revenue.nsmallest(3)

print('Top 3 categories with the highest revenue:')
print(top_3)
print('Bottom 3 categories with the lowest revenue:')
print(bottom_3)
```

**Q9.** Group the dataset by Region and calculate the total revenue generated in each region. Identify the
region with the highest revenue and the region with the lowest revenue.

```python
region_revenue = df.groupby("Region")["Revenue"].sum()

highest_revenue_region = region_revenue.idxmax()
lowest_revenue_region = region_revenue.idxmin()

print('Region with the highest revenue:')
print(highest_revenue_region)
print('Region with the lowest revenue:')
print(lowest_revenue_region)
```

**Q10.** Extract the month from the Date column using Pandas and group the dataset by month to calculate the
total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column
using NumPy.

```python
df["Month"] = pd.to_datetime(df["Date"])
df["Month"] = df["Month"].dt.month

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print('Total monthly revenue:')
print(monthly_revenue)

mean_revenue = np.mean(df["Revenue"])
median_revenue = np.median(df["Revenue"])
std_dev_revenue = np.std(df["Revenue"])

print('Mean revenue:')
print(mean_revenue)
print('Median revenue:')
print(median_revenue)
print('Standard deviation of revenue:')
print(std_dev_revenue)
```
