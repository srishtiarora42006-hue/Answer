**Q6.** Explain the difference between the following three sharing methods in Power BI Service: Direct Report Sharing, Workspace Access, Power BI Apps
Also mention one use case for each.

->
*  Direct Report sharing - this method of sharing is used when specific reports or dashboards need to be shared with a selected group of people or any one of them, this allows the user to directly share the particular document to its user . Use case - Report regarding the sale of goods in the month of june being shared to sales manager through direct sharing.

* Workspace Access - this method is used when the report which needs to be shared is required to be sent to the entire workspace depending on what their role is , it provides the person an access to all the reports and dashboards inside the folder rather than just one singular report, further the users of reports with workspace access can edit, design,calculate any field they want.
Use case - A Sales Report being shared to the sales department by allowing workspace access so now they can get, see and edit all the documents,dashboards,tables etc present in that folder .

* Power BI apps - Through this method the reports, dashboards from power BI can be shared to many users with read only access, this is also a way of communication by sharing of reports online.
Use case- Reports regarding the IPO  of the company being shared to the employees accross organisations.

**Q7.** Describe the four workspace roles available in Power BI and explain how access control is managed using these roles.


-> The workspace roles available in power BI are :-

* ADMIN - An admin has full control over the workpsace and its settings, control of access and permissions , an admin can edit,publish,create,design,create or even delete content from the dataset.

* MEMEBER- like an admin a member can also create,publish,share and edit reports but he/she doesnot have all the access and permissions just like an admin does, i.e only some of the settings are accesisble to the member.

* CONTRIBUTOR- a contributor can also do all the tasks of creating,sharing,publishing,editing of the reports but does not have any access to workspace setting, controls and permissions.

* VIEWER- A viewer has only read only access , can interact with the reports and dashboards but does not have any permissions to create,edit or publish any content.

**Q9.** Explain Row-Level Security (RLS) in Power BI.
How does RLS ensure that different users see different data using the same report?

-> Row level security is afeautre in Power BI where the data is restricted and accessible to the users at row level, which means specific people can read or go through specific records and not all of the report, this is done by logging in roles in power BI while publishing and using a DAX expression which creates a specific constraint for specific type of user. RLS ensures diffrent users see diffrent data by verifying the role assigned to that user when he/she opens the report, it automatically shows only those records which are accessible to the users.
