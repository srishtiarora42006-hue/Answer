**Question 5** : Convert the following inconsistent “Gender” entries into a standardized format (“Male”, “Female”):["M", "male", "F", "Female", "MALE", "f"]



```python
import pandas as pd
gender = ['M', 'male', 'F', 'Female', 'MALE', 'f']
standardized = []
for g in gender:
  g = g.lower()
  if g == 'm' or g == 'male':
    standardized.append('Male')
  elif g == 'f' or g == 'female':
    standardized.append('Female')
  else:
    standardized.append(g)
df = pd.DataFrame( {
    'original': gender,
    'standardized' : standardized
    })
print(df)
```

**Question 6** : What is One-Hot Encoding? Give an example with the categories: “Red, Blue, Green”.

-> one hot encoding creates a binary column for category, it indiviually shows each category and may also increase the column count. It is consisdered the best for nominal categories.




```python
import pandas as pd
data = {
  'colour' : ['Red', 'Blue', 'Green'] ,
  'Red' : [1, 0, 0],
  'Blue' : [0, 1, 0],
  'Green' : [0, 0, 1]
}
df = pd.DataFrame(data)
df_encoded = pd.get_dummies(df, columns=['colour'])
print(df_encoded)
```
