```python
import pandas as pd
```

```python
df = pd.read_excel("/content/employee_productivity_dataset.xlsx")
print(df)
```

Assignment Questions

**Q1**. Handle Missing Values (Basic Cleaning)

Fill missing values using the following:

Age → Median

Salary → Mean

Hours_Worked_Per_Week → Median

Performance_Score → Mean

Display the dataset after handling missing values.

```python
#first we will detect the number of missing values per columns to find out how many values in the datset will be imputed
df.isna().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(df["Hours_Worked_Per_Week"].median())
df["Performance_Score"] = df["Performance_Score"].fillna(df["Performance_Score"].mean())
```

```python
#dataset after handling missing values
print(df)
```

**Q2.** Label Encoding

Convert these columns into numeric values using Label Encoding:

Gender

Department

Show the updated columns.

```python
from sklearn.preprocessing import LabelEncoder
Encoder = LabelEncoder()
df["Gender_Encoded"] = Encoder.fit_transform(df["Gender"])
df["Department_Encoded"]= Encoder.fit_transform(df["Department"])
print(df[["Gender_Encoded","Department_Encoded"]])
```

**Q4.** Normalization (Min-Max Scaling)

Normalize the following columns:

Salary

Hours_Worked_Per_Week

Use MinMaxScaler and display results.

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[["Salary_Normalised","Hours_Worked_Per_Week_Normalised"]] = scaler.fit_transform(df[["Salary","Hours_Worked_Per_Week"]])
print(df)
```

**Q5.** Standardization (Scaling)

Apply StandardScaler on:

Age

Projects_Completed

Display the transformed values.

```python
from sklearn.preprocessing import StandardScaler
Scaler = StandardScaler()
df["Age_Standardised"] = Scaler.fit_transform(df[["Age"]])
df["Projects_Completed_Standardised"] = Scaler.fit_transform(df[["Projects_Completed"]])
print(df[["Age_Standardised","Projects_Completed_Standardised"]])
```

**Q7**. Build Preprocessing Pipeline

Create a pipeline that:

Applies encoding to categorical columns

Applies scaling to numerical columns

Use:

ColumnTransformer

Pipeline

```python
from sklearn import pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

categorical_columns = ["Name","Gender","Work_Mode","Location","Department"]
Numerical_columns = ["Age","Hours_Worked_Per_Week","Projects_Completed","Salary","Performance_Score" ]
preprocessor = ColumnTransformer(
    transformers=[
        ("cat",OneHotEncoder(),categorical_columns),
        ("num",MinMaxScaler(),Numerical_columns)
    ]
)
pipeline  = Pipeline([
    ("preprocessor",preprocessor)
])

```

**Q8.** Apply Pipeline

Apply the pipeline on the dataset.

Display:

Transformed dataset

Shape of final dataset

```python
preprocessed_data = pipeline.fit_transform(df)
print(preprocessed_data)
#shape of the final dataset can be known by -
print(preprocessed_data.shape)
```
