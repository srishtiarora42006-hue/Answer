  2. how would you sort the following dataset first by "Departmen"(A-Z) then "salary" ( largest to smallest) create a step by step approach.





        


```python
import pandas as pd
#create dataset
data = {"Employee" : ["Sonu","Panav","Rahul"],"Department": ["IT","HR","IT"],"Salary": [4000,5000,2500]}
df = pd.DataFrame(data)
#sort by department (A-Z) and Salary (descending)
sorted_df= df.sort_values(by=["Department", "Salary"], ascending=[True, False])
sorted_df
```

### step-by-step approach:
1. created the data set using pandas DataFrame.
2. used sort_values() function,
3. sorted Department in ascending order(A-Z).
4. sorted Salary in descending order (Largest to smallest).
5. displayed the final sorted dataset.
