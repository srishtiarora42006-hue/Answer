**Question 1** : What is Data Loading in Power BI and why is it considered the first step of analysis?

-> Data loading in power BI is considered the first step as data needs to be imported from somewhere for its analysis to be performed to derive any meaningful conclusions, once the data is imported it can be cleaned and transformed and thus be analysis ready.

**Question 2** : Explain the difference between “Load” and “Transform Data” in Power BI.

-> To **load data** means to import data from various data sources into the power BI desktop application for performing any analysis and making any reports, whereas  to **transform data** refers to clean, modify correct the data by making it consistent and standardized so that the further analysis from this data can be useful and accurate.


**Question 3** : What is a Fact Table and a Dimension Table? Give examples from the dataset.

-> **Fact Table** - these are the types of tables which store transactional or numerical data in them, that means it has  all the columns which have digits as their values.

**Dimension Table**- these tables store such values which are categorical, non numerical or descriptive data , it contains all the columns with are non transactional in nature.


**Question 4** : Why is Star Schema preferred over Snowflake Schema in Power BI?

-> Star schema is simple data modeling design in which a fact table is connected to various dimension tables forming a star-shape like structure. Star schema shows all the relationships very clearly for easy understanding and also stimulates the power query performance. hence, it is preferred over snowflake schema.
