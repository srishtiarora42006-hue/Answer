**1.** What is the difference between a Basic Filter, Advanced Filter, and a Slicer in Power BI?

-> Filters are used in power BI to view custom data fields , these enable the user to view only those particular values which are more significant from the user's point of view and is filtered from the rest of the data.

**Basic Filters** - These are the simplest form of filters used in power BI , they enable the user to  manually select some values which needs to be preseneted, usually basic filters are used when one is know of what to keep in the data, FOR EXAMPLE - to show sales for the month of january only.

**Advanced Filters** - These filters provide the user with advanced filtering options , unlike basic filters they also allow the user to do multi condition filtering , through advanced filters one can do complex and dynamic filtering which helps to present data with more precision.
FOR EXAMPLE - to show sales occured in Dubai for the year 2025 .

**Slicer** - Slicers are button type of visual which filter the data on the spot on clicking that particular data field, they are very interactive which also enhance representation as well as are more functional. one just needs to click on a particular field and data gets sorted according to that particular legends, FOR Example - if slicer is added and field added is product, with values as - sum of sales , so when one clicks a particular product the data visual will show sum of sales for that particular product .

**4**. What is a Visual-Level Filter, and how does it differ from Page-Level and Report-Level Filters?

-> **visual level filter** only applies to a particular selected filter , in the filtering pane when we filter a particular field the changes show up in that particular visual, which represents a visual level filter.

**page level filter** this type of filter is the visible after the visual level filter in the filters pane, these filters apply to all the visuals present at that particular page, hence the action of filtering is applied to all the visuals prsent on that particualr page .

**report level filter** these filters are applied on the whole report irrespective of page , once the user drags and drops down a field in the report level filter in the filters pane, the filtering action gets executed to whole of the report, this is very useful if one needs to filter entire report at once and also saves a lot of time.
