Question 5 : Explain three common challenges faced during data extraction.

-> During the process of extraction of data a number of challenges can occur however the most common are ▶

1.  **Data may contain missing values** - this is one of the most common data extraction problem that data may have some missing values which ultimately may lead lesser data reliability or improper analysis.

2.  **Scattered data**  - the second most common is that the data is scattered and needs to be gathered from a lot of sources which later needs to be integrated with various other datasets to perform an analysis.

3. **wrong data may lead to wrong analysis** - it is also very important that data which needs to be extracted should be correct as extraction of wrong data may lead to wrong analyis , hence providing biased or inconsistent results which are not reliable.
