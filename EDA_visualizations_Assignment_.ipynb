**Q1.** Create a countplot to show the number of passengers by gender

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x = "Sex", data=df)
```

INTERPRETETION

-The above bar graph shows the distribution of males and females in the dataset. It can be easily inferred from the above dataset that there were larger number of males i.e > 650 in comparison to females i.e 300 approx.

**Q2.** Create a bar chart to show the number of passengers in each passenger class using different colors.

```python
df["Pclass"].value_counts().sort_index().plot(kind="bar", color=["red", "green", "blue"])
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Number of Passengers in Each Passenger Class")
plt.show()
```

INTERPRETATION-

According to the above bar graph it is evident that most number of Passengers belong to passenger class 3 followed by 1 and then 2 , blue ,red and green coloured bars represents each of them repectively.

**Q3.** Create a histogram to show the distribution of passenger age.

```python
plt.hist(df["Age"])
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Passenger Age")
plt.show()
```

INTERPRETATION -

The above histogram shows that the maximum passengers belong from the age group 20 to 30 which ranges from 150-350 passangers.

**Q4.** Create a boxplot to detect outliers in the fare column.

```python
import seaborn as sns
sns.boxplot(x=df["Fare"])
plt.title("Boxplot of total Fare")
plt.show
```

INTERPRETATION-

The above box plot shows us that there are many outliers present in the fare column, the blue box represent middle 50% of the values, the line inside it reprents the median and the line on either side of the box shows the range of minimum to maximum values and all the circles outside represent the outliers.


**Q5.** Create a scatterplot to

 a. show the relationship between age and fare.

 b. analyze the relationship between family size (SibSp + Parch) and fare.

```python
# relationship between age and fare
sns.scatterplot(x="Age",y="Fare",data=df)
plt.title("Relationship between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

INTERPRETATION

The above scatter plot shows the relationship between age of passengers and fare paid by them, there is a huge number of people which belonged to age group between 10-50, and fare prices seems to be almost around the same range, however there re ceratin cases where age is less and fare is more, hence, this scatter plot does not represent a positive correlation overall.

```python
#relationship between family size and fare
sns.scatterplot(x="SibSp",y="Fare",data=df)
plt.title("Relationship between Family size and Fare")
plt.xlabel("Family size")
plt.ylabel("Fare")
plt.show()
```

INTERPRETATION -

The scatter plot shows the relationship between family size and fare. Most passengers had a smaller family size (between 0 and 3), with fares varying widely. A few passengers with small family sizes paid very high fares, while passengers with larger family sizes generally paid comparatively lower fares.  there does not appears to be a strong positive relationship between family size and fare.

**Q6**. Create a countplot to show survival count by gender.

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

INTERPRETATION -

The above countplot shows the survived count distribution as per the gender i.e male and female, the blue colour represents the count of people who didnot survive and orange represents the number of people who did.

**Q7**. Create a bar chart to show survival by passenger class.


```python
df.groupby("Pclass")["Survived"].sum().sort_index().plot(kind="bar", color=["red", "green", "blue"])
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Number of Passengers who survived in Each Passenger Class")
plt.show()
```

INTERPRETATION-

The above bar graph represents the number of passengers in the 3 passenger classes, i.e 1,2 and 3. From the graph it is evident that passenger class 1 has maximum number of passengers and passengers class 3 has least number of passengers.

**Q8.** Create a heatmap to show correlation between numerical variables

```python
corr = df.corr(numeric_only= True)
sns.heatmap(corr,annot = True)
plt.title("Correlation Heatmap")
plt.show()
```

INTERPRETATION

The above heatmap shows the correlation between all the numerical data present in the dataset which includes - Passenger ID , Survived count, Passenger class,Age,SibSp,Parch and fare, as the colour gets darker it represents a weak or negative correlation and the colour which goes towards orange shows strong or positive correlation.
