2. Why are some views updatable while others are read-only? Explain with an example.

 -> view is the outermost layer which is visible to the end users or clients , it is basically a virtual window which contains the info the user has asked for . An updateable view is the one in which the values which are already logged into the data can be altered/modified , whereas  a read only view provides the end user to just have a look at the data and one cannot make any alterations to it. A view is updateable if it directly shows rows from one table , whereas a view can be read only if the rows cannot be determined into a single row. FOR  example - views using joins, aggregations are read only , whereas single row value rows are updateable.
