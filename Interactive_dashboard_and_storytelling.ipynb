**Q1**. Explain the diffrence betweena Tableau dashboard and Tableau Storypoint. give 2 healthcare related use cases.  

-> Dashboards are a collection of various sheets which refelct various insights in a group and present important KPIs, Dashboards allows the users to use filters, drill downs etc , it helps to analyse all the significant points at one glance and provide a summarised view to users of the dashboards.

Tableau story points are a sequence of sheets or dashboards which help to explain a story like view of the data or its narrative, various sheets are used indiviually in a particular order which are based on the same point and show various aspects revolving around the same topic to present it as a story.

**Q3.**  Patient distribution

Create a bar chart showing:
* Total patients by hospital region.
* Add this as a story point with proper title.

Steps for the following

▶ 1. Drag the hospital region to columns.

2. Drag the patient ID field to Rows.

3. click on the dropdown in the patient ID field in the rows section and select Count measure.

4. go to marks and change the type to bar.

5. Sort the bars in descending order by clicking on the sort icon next to the header.

6. turn on the labels.

7. Hold ctrl and drag the CNT(PatientID) from rows to colour to see colour grading as per count of patient id.

8. Format the text as per desired.

**INSIGHTS DERIVED**

* Minimum number of patients come from the centeral region.
