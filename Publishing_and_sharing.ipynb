**1**. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/Cloud in terms of accessibility and data security?

-> **Tableau public** allows a user to publish a dashboard which is accesible to whole world, without any security or confidentiality , it is designed for sharing the visualisations to n number of people using internet, but there is no security for confidential reports or dashboards, Therefore it is suitable for common information which is non-sensitive and needs to be shared to numerous people.

**Tableau Server/cloud**- When a user publishes  a dashboard or reports to these platforms , they require a proper sign in to authenticate the user to give him/her access to the information based on their roles and permissions , hence it is more suitable for sensitive and user based or role based information which needs to be specifically shared , it allows controlled access and higher level of security.

How does this affect data refresh and performance?

-> When a data source is connected to a live connection the data gets updated time to time, it means we get real time data and dashboards which are made from such visualisations which are connected to a live data source connection gets updated from time to time whenever the user opens it, on the other hand when a data source is connected to an extract connection then the data being showed in the dashboard is just the original data which does not get timely updated but also offers a faster performance as it need not read new data and hence optimizes faster than a live connection.

**4.** If you want users to view a dashboard but not download the underlying data, which permission setting would you adjust in Tableau Server?

-> In tableau to prevent the users from download the underlying data but allow them to view the dashboard , the adminstrator should disable download data permission, this permission allows the user to view as well as download the dashboard and its data , when this is disabled users will still be able to view all the data in the dashboard and interact with it but they will not be able to download it.

**6.**  List two best practices you should follow when preparing a dashboard for public sharing to ensure data security and privacy.

-> The best practices which can be followed when preparing a dashboard for public sharing to ensure data security and privacy are:

* Remove sensitive data - Before publishing the dashboards and reports make sure all the confidential information is either removed or masked to protect the security of the data and avoid unaurthorised access.

* Review permissions assigned - Ensure that all the permissions which are being provided for are accurate, role based access is correct to prevent data leaks , ensure permission for downloading the dashboard and its underlying data is disabled (if required).
