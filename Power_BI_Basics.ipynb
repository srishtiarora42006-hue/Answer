1.  What is Power BI and why is it used in businesses?

-> Power BI is a business intelligence tool by microsoft which offers to load data from various sources and then formulate reports from the data for their further analysis and decision making. It is used in business as it helps to form reports regarding the business performance, business insights which the helps the analysts and decision makers to take further steps.


2. Name and explain the three main components of Power BI.

-> The three main components of Power BI are

▶
 * **Power BI desktop** - this is free desktop application which helps to clean data, transform it and then make reports for useful business insights.

  * **Power BI service** - it is a cloud based platform used to publish and share the reports made and connect to other users of such reports and analysis.

  * **Power BI mobile app** - it is a mobile application which is used to view the reports and interact with them , it is accesible on a mobile phone.  

3. Explain the Power BI workflow.

-> The power BI workflow is distributed into 4 main steps which are explained as below:

* **CONNECT** - This is the first step , it refers to connecting to various sources through which data needs to be loaded, Power BI provides many sources to get data from such as SQL, CSV, excel, etc.

* **TRANSFORM** - The second step is to transform the data , it can be filtering , sorting, data standardization, integration of data, changing data types etc, this step ensures that data is now fit for intended use.

* **VISUALIZE** - Next step is to visualize the transformed data, visualisation refers to formulation of reports and analyse business patterns and trends which are further beneficial for business.

* **SHARE** - The last step of the workflow is to share the reports made with the users of such reports , for eg. analysts, decision makers, enterprise heads etc.
