**Question 4** : Differentiate between exact, partial, and fuzzy duplicates.

-> **Exact duplicates** - it refers to when the records in the column are completely identical, that means there is no diffrence between them, E.G - same name, same email, same phone number, etc.

**Partial duplicates** - partial duplicates refer to a sitaution when the key business entity matches but rest of the columns are diffrent, E.G - same phone number but diffrent name, or same email ID but diffrent address.

**Fuzzy dupliactes** - it refers to when value is refering to the same entity but there is slight variation, this usually happens due to lack of standardization E.G - Bengluru and Banglore.
