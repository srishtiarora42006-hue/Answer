Q4. Why is median imputation preferred over mean imputation for skewed data such as income?

-> Mean imputation is sensitive to the presence of outliers which may lead to a biased result whereas median imputation is not affected by the presence of outliers, when we have skewed or very distinct and diverse data , it is possible that results are biased result when mean is used as a measure for imputation which can be easily avoided by using median.
