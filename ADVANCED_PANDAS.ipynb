**Q3** Customer Purchase Analysis

Identify the top 5 customers based on total sales.

**Q4** Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?

```python
Regional_sales = df.groupby("Region")['Sales'].sum()
```

```python
print(Regional_sales)
```

```python
Highest_revenue = Regional_sales.idxmax()
print("The region with the highest revenue is:", Highest_revenue)
```

**Q5** Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?

```python
Total_quantity_sold = df.groupby("Product")['Quantity'].sum()
print(Total_quantity_sold)
```

```python
Most_popular_product = Total_quantity_sold.idxmax()
print("The most popular product is:", Most_popular_product)
```

**Q6**Multi-Level Grouping

Group the dataset by: Region and Category

Calculate the total sales for each combination.

What insights can you derive from this analysis?

```python
#group by region and category
df.groupby("Region")["Category"].value_counts()
```

```python
#total sales for each category
total_sales = df.groupby(["Region", "Category"])["Sales"].sum()
print(total_sales)
```

Insights derived from the above analysis are:-
- The maximum sales of accessories and Electronics were in the East region.
- The maximum sales of office supplies were in North region.
